# Camera Geometry for NeRF

This notebook explores the camera geometry used in NeRF:
- Pinhole camera model
- Camera intrinsics and extrinsics
- Camera-to-world transformation matrices
- Ray generation from pixel coordinates

```python
import numpy as np
import matplotlib.pyplot as plt
import torch

# --- Pinhole camera: pixel -> ray direction ---
def get_rays(H, W, focal, c2w):
    """Generate rays for every pixel given a camera-to-world matrix."""
    i, j = np.meshgrid(np.arange(W, dtype=np.float32),
                        np.arange(H, dtype=np.float32), indexing='xy')
    dirs = np.stack([(i - W * 0.5) / focal,
                     -(j - H * 0.5) / focal,
                     -np.ones_like(i)], axis=-1)  # (H, W, 3)
    # Rotate ray directions from camera to world frame
    rays_d = (dirs[..., np.newaxis, :] * c2w[:3, :3]).sum(axis=-1)
    rays_o = np.broadcast_to(c2w[:3, 3], rays_d.shape)
    return rays_o, rays_d

# Example: identity pose, 100x100 image, focal=100
H, W, focal = 100, 100, 100.0
c2w = np.eye(4, dtype=np.float32)
rays_o, rays_d = get_rays(H, W, focal, c2w)
print('Ray origins shape:', rays_o.shape)
print('Ray directions shape:', rays_d.shape)
```

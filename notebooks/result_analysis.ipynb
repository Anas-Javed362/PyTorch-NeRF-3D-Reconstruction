# NeRF Training Result Analysis

This notebook analyses and visualises NeRF training results:
- PSNR / SSIM curves over training iterations
- Comparison across ablation experiments
- Side-by-side rendered vs. ground-truth image grids
- Per-scene quantitative summary table

```python
import json
import glob
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

METRICS_DIR = Path('../outputs/metrics')

def load_metrics(metrics_dir: Path):
    """Load all metrics JSON files from a directory into a dict keyed by run name."""
    results = {}
    for path in sorted(metrics_dir.glob('*.json')):
        with open(path) as f:
            results[path.stem] = json.load(f)
    return results

# --- Synthetic data for demonstration ---
iters = np.arange(0, 50001, 500)
psnr  = 10 + 15 * (1 - np.exp(-iters / 15000)) + np.random.randn(len(iters)) * 0.3

plt.figure(figsize=(8, 4))
plt.plot(iters, psnr, label='lego_baseline')
plt.xlabel('Iteration')
plt.ylabel('PSNR (dB)')
plt.title('Training PSNR curve')
plt.legend()
plt.tight_layout()
plt.show()

print(f'Peak PSNR: {psnr.max():.2f} dB at iter {iters[psnr.argmax()]}')

# When real metrics are available, replace the synthetic data above with:
# results = load_metrics(METRICS_DIR)
# for run_name, data in results.items():
#     plt.plot(data['iters'], data['psnr'], label=run_name)
# plt.legend(); plt.show()
```

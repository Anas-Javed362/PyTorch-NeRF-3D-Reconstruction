# Ray Visualization for NeRF

This notebook visualizes the rays cast through a scene:
- 3D ray bundles from multiple camera poses
- Near/far sampling bounds
- Stratified sample point positions along rays
- Scene coverage analysis

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# --- Stratified sampling along a ray ---
def sample_along_ray(near, far, num_samples, perturb=True):
    """Sample points between near and far along a ray."""
    t_vals = np.linspace(near, far, num_samples)
    if perturb:
        mids = 0.5 * (t_vals[1:] + t_vals[:-1])
        lower = np.concatenate([t_vals[:1], mids])
        upper = np.concatenate([mids, t_vals[-1:]])
        t_vals = lower + (upper - lower) * np.random.rand(*t_vals.shape)
    return t_vals

# Visualize sample points along a single ray
near, far, num_samples = 2.0, 6.0, 64
t_vals = sample_along_ray(near, far, num_samples, perturb=True)

ray_origin = np.array([0.0, 0.0, 0.0])
ray_dir    = np.array([0.0, 0.0, -1.0])
points = ray_origin[np.newaxis] + t_vals[:, np.newaxis] * ray_dir[np.newaxis]

fig = plt.figure(figsize=(10, 4))
ax = fig.add_subplot(121, projection='3d')
ax.scatter(points[:, 0], points[:, 1], points[:, 2], c='steelblue', s=10)
ax.set_title('Sample points along ray')
ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')

ax2 = fig.add_subplot(122)
ax2.eventplot(t_vals, orientation='horizontal', linewidths=1)
ax2.set_title('t-value distribution (stratified)')
ax2.set_xlabel('t'); ax2.set_yticks([])
plt.tight_layout()
plt.show()
print(f'Sampled {len(t_vals)} points between t={near} and t={far}')
```
